```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('/content/evgeni-tcherkasski-c659bBmJpw0-unsplash.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

# Global Threshold

```python
_, th_binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
_, th_binary_inv = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY_INV)
_, th_trunc = cv2.threshold(gray, 127, 255, cv2.THRESH_TRUNC)
_, th_tozero = cv2.threshold(gray, 127, 255, cv2.THRESH_TOZERO)
```

# Adaptive Threshold

```python
th_adaptive_mean = cv2.adaptiveThreshold(
    gray, 255,
    cv2.ADAPTIVE_THRESH_MEAN_C,
    cv2.THRESH_BINARY,
    11, 2)

th_adaptive_gaussian = cv2.adaptiveThreshold(
    gray, 255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    11, 2)
```

# Otsu Threshold

```python
_, th_otsu = cv2.threshold(
    gray, 0, 255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU)
```

```python
titles = [
    'Original',
    'Binary',
    'Binary Inv',
    'Trunc',
    'ToZero',
    'Adaptive Mean',
    'Adaptive Gaussian',
    "Otsu"
]

images = [
    gray,
    th_binary,
    th_binary_inv,
    th_trunc,
    th_tozero,
    th_adaptive_mean,
    th_adaptive_gaussian,
    th_otsu
]

plt.figure(figsize=(15, 8))

for i in range(len(images)):
    plt.subplot(2, 4, i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```
